```python
from matplotlib import pyplot as plt
import os

folder_path = '/Users/rl884/Downloads/PICIN/Result/0310/'

Result_Flow = []
for file_num in range(1, 31):
    inputfile = open(folder_path+'Test{}_noChIP_0310_ResultFlow.txt'.format(file_num))
    Result_Flow.append([])
    for line in inputfile:
        Result_Flow[-1].append(line.split())
    inputfile.close()
```

```python
fig = plt.figure(figsize=(10,10))
for i in range(0, len(Result_Flow)):
    temp_to_plot = []
    for j in range(0, len(Result_Flow[i])):
        temp_to_plot.append(float(Result_Flow[i][j][-1]))
    plt.plot(temp_to_plot, linestyle='dotted', c='black')
    #print(temp_to_plot)
plt.ylim(0,20)
plt.xlabel('Iterations', fontsize=20)
plt.ylabel('Objective Function Readout', fontsize=20)
plt.show()
```
